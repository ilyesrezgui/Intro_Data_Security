```python
import io
import pandas as pd

# Load only the first 50,000 rows for performance
df = pd.read_csv(io.BytesIO(uploaded['weather_data.csv']), nrows=50000)

# Convert to datetime and sort for time series integrity
df['Date_Time'] = pd.to_datetime(df['Date_Time'])
df = df.sort_values(by='Date_Time')

# Preview the result
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Select only 4 numeric columns
features = ['Temperature_C', 'Humidity_pct', 'Precipitation_mm', 'Wind_Speed_kmh']
df_small = df[features].copy()

# Normalize the features to [0, 1]
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df_small)

# Turn into time series sequences (24 time steps each)
sequence_length = 24
sequences = [scaled_data[i:i+sequence_length] for i in range(len(scaled_data) - sequence_length)]
sequences = np.array(sequences)

# Show shape (should be [samples, 24, 4])
print("✅ Final shape:", sequences.shape)
```

```python
# Minimal working TimeGAN setup (simplified)
import tensorflow as tf
from tensorflow.keras import layers

class SimpleTimeGAN(tf.keras.Model):
    def __init__(self, seq_len, feature_dim):
        super(SimpleTimeGAN, self).__init__()
        self.seq_len = seq_len
        self.feature_dim = feature_dim

        self.generator = tf.keras.Sequential([
            layers.Input(shape=(seq_len, feature_dim)),
            layers.LSTM(100, return_sequences=True),
            layers.TimeDistributed(layers.Dense(feature_dim, activation='sigmoid'))
        ])

    def call(self, inputs):
        return self.generator(inputs)
```

```python
import tensorflow as tf
from tensorflow.keras import layers

# Data shape
seq_len = sequences.shape[1]
feature_dim = sequences.shape[2]

# Generator Model
def build_generator():
    model = tf.keras.Sequential([
        layers.Input(shape=(seq_len, feature_dim)),
        layers.LSTM(100, return_sequences=True),
        layers.TimeDistributed(layers.Dense(feature_dim, activation='sigmoid'))
    ])
    return model

generator = build_generator()

# Compile and train
generator.compile(optimizer='adam', loss='mse')
generator.fit(sequences, sequences, epochs=10, batch_size=64)

# Generate synthetic data
noise = tf.random.normal((1, seq_len, feature_dim))
synthetic = generator.predict(noise)[0]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(sequences[0])
plt.title("Real Sequence")

plt.subplot(1, 2, 2)
plt.plot(synthetic)
plt.title("Synthetic Sequence")

plt.tight_layout()
plt.show()
```
